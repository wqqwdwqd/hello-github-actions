## Основы языка сценариев AWK

### Пример 1: Базовые операции с полями

```python
!cat src/myfile
```

```python
!gawk '{print $1}' src/myfile
```

```python
!gawk '{print $3}' src/myfile
```

### Пример 2: Изменение регистра символов

```python
!echo -e "This line contains bar.\nThis line contains foo." | gawk '/foo/ { print toupper($0); }'
```

### Пример 3: Вывод фамилий, имен и оценок

```python
!gawk '{print $1, $2, $5}' src/list_students.txt
```

### Пример 4: Поиск студентов с оценкой 8

```python
!gawk '/8/ {print $0}' src/list_students.txt
```

### Пример 5: Студенты с оценкой 4

```python
!gawk '($5==4)' src/list_students.txt
```

### Пример 6: Студенты специальности КБ, 1 курса

```python
!gawk '($3=="КБ")&&($4==1)' src/list_students.txt
```

### Пример 7: Имена длиной >5 символов

```python
!gawk 'length($2) > 5' src/list_students.txt
```

### Пример 8: Нумерация строк

```python
!gawk '{print NR, $0}' src/list_students.txt
```

### Пример 9: Имена в верхнем регистре

```python
!gawk '{print toupper($2)}' src/list_students.txt
```

### Пример 10: Суммарный балл оценок

```python
!gawk '{sum += $5} END {print("SUM=", sum)}' src/list_students.txt
```

### Пример 11: Поиск фамилии Кравченя

```python
!gawk '($1 ~ /Кравченя/)' src/list_students.txt
```

### Пример 12: Поиск с регистронезависимостью

```python
!gawk '($1 ~ /[Лл]ипин/)' src/list_students.txt
```

```python
# Все записи КРОМЕ Липин
!gawk '($1 !~ /[Лл]ипин/)' src/list_students.txt
```
