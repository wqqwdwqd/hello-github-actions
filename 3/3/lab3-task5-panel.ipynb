```python
import panel as pn
import pandas as pd
import numpy as np
import plotly.express as px

# Правильно включаем расширения
pn.extension('plotly')

print("Panel успешно загружен с поддержкой Plotly!")

# Создаём тестовые данные
np.random.seed(42)
data = pd.DataFrame({
    'x': np.random.randn(50),
    'y': np.random.randn(50),
    'category': np.random.choice(['A', 'B', 'C'], 50),
    'value': np.random.randint(10, 100, 50)
})

# Показываем данные
print("Пример данных:")
print(data.head())
```

```python
# Создаём простой график Plotly
fig = px.scatter(data, x='x', y='y', color='category', 
                 title='Простая диаграмма рассеяния')
plot_pane = pn.pane.Plotly(fig, height=400)
plot_pane
```
