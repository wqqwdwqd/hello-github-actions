# Упражнение 4 — Разные стили линий

```python
# Создаём массив значений по оси X от -5 до 5 (100 точек)
x = np.linspace(-5, 5, 100)

# Вычисляем значения синуса для каждого X
y = np.sin(x)

# Задаём размер области графика (ширина=10, высота=6 дюймов)
plt.figure(figsize=(10, 6))

# Строим 4 линии с разной толщиной (lw) и стилем (ls)
plt.plot(x, y, lw=1, ls='-', label='Сплошная')           # обычная линия
plt.plot(x, np.sin(x+0.5), lw=2, ls='--', label='Пунктирная')  # пунктир
plt.plot(x, np.sin(x+1), lw=3, ls='-.', label='Точка-тире')    # точка-тире
plt.plot(x, np.sin(x+1.5), lw=4, ls=':', label='Точка-точка')  # точка-точка

# Добавляем легенду (подписи линий)
plt.legend()

# Устанавливаем заголовок графика
plt.title('Разные стили линий', fontsize=17)

# Включаем сетку для удобства восприятия
plt.grid()

# Сохраняем график в файл с высоким качеством (300 dpi)
plt.savefig('task7_exercise4.jpeg', dpi=300)

# Показываем график на экране
plt.show()
```

# Упражнение 5 — Маркеры

```python
# Маркеры выделяют точки на линии
x = np.arange(7)
y1 = np.array([3, 1, 2, 4, 5, 3, 4])
y2 = np.array([5, 4, 5, 3, 4, 2, 5])

plt.figure(figsize=(9, 6))
plt.plot(x, y1, marker='o', markersize=20, label='Петя')  # круги
plt.plot(x, y2, marker='^', markersize=10, label='Вася')  # треугольники
plt.legend()
plt.grid()
plt.savefig('task7_exercise5.png', dpi=300)
plt.show()
```

# Упражнение 6 — Изменение цвета линий и маркеров

```python
# Можно задать цвета линий и точек отдельно
plt.figure(figsize=(9, 6))
plt.plot(x, y1, marker='o', markersize=20, color='blue',
         markerfacecolor='lightblue', markeredgecolor='navy', label='Петя')
plt.plot(x, y2, marker='^', markersize=10, color='red',
         markerfacecolor='pink', markeredgecolor='darkred', label='Вася')
plt.legend()
plt.grid()
plt.savefig('task7_exercise6.jpeg', dpi=300)
plt.show()
```

# Упражнение 8 — Разделение данных в Seaborn (tips)

```python
# Загружаем встроенный набор данных о чаевых в ресторане
tips = sns.load_dataset('tips')

# Строим диаграмму: зависимость чаевых от суммы счёта
sns.relplot(x='total_bill', y='tip', hue='sex', col='time', data=tips)
plt.savefig('task7_exercise8.jpeg', dpi=300)
plt.show()
```

# Упражнение 9 — Линейный график на основе набора данных fmri

(демонстрация временных рядов)

```python
fmri = sns.load_dataset('fmri')

# Создаем plot с помощью seaborn
plot = sns.relplot(
    data=fmri,
    x='timepoint',
    y='signal',
    kind='line',
    hue='event',  # Исправлено hues -> hue
    style='region',
    errorbar=None  # Исправлено ci -> errorbar
)

# Устанавливаем заголовок и сохраняем
plot.fig.suptitle('Активность мозга по временным рядам (FMRI)', fontsize=15)
plot.savefig('task7_exercise9.jpeg', dpi=90)  # Исправлено quality на dpi
plt.show()
```

# Упражнение 10 — Гистограмма распределения признака

```python
# Загружаем данные о пингвинах
penguins = sns.load_dataset('penguins')

# Гистограмма + линия плотности
sns.histplot(data=penguins, x='flipper_length_mm', bins=20,
             kde=True, color='skyblue')
plt.title('Распределение длины ласт', fontsize=15)
plt.savefig('task7_exercise10.png', dpi=300)
plt.show()
```

# Упражнение 11 — Диаграмма распределения (displot)

```python
# Загружаем данные пингвинов
penguins = sns.load_dataset('penguins')

# Создаем диаграмму распределения
plot = sns.displot(
    data=penguins,
    x='flipper_length_mm',
    hue='species',
    col='sex',
    kind='kde',
    fill=True
)

# Сохраняем изображение
plot.savefig('task7_exercise11.jpeg', dpi=85)  # Исправлено quality на dpi
plt.show()
```

# Упражнение 12 — Ящиковая диаграмма (boxplot)

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=penguins, x="species", y="body_mass_g", hue="species", palette='Set2', legend=False)
plt.title('Масса тела у разных видов пингвинов', fontsize=15)
plt.xlabel('Вид пингвина')
plt.ylabel('Масса тела (г)')
plt.savefig('task7_exercise12.png', dpi=300, bbox_inches='tight')  # Исправлено имя файла
plt.show()
```

# Упражнение 13 — Violinplot (диаграмма распределения с плотностью)

```python
plt.figure(figsize=(10, 6))
sns.violinplot(data=penguins, x='species', y='flipper_length_mm', hue='species', palette='Pastel1', legend=False)
plt.title('Распределение длины ласт у пингвинов', fontsize=15)
plt.xlabel('Вид пингвина')
plt.ylabel('Длина ласт (мм)')
plt.savefig('task7_exercise13.jpeg', dpi=85, bbox_inches='tight')  # Исправлено quality на dpi
plt.show()
```

# Упражнение 14 — Тепловая карта корреляции

```python
# Считаем корреляции между числовыми признаками
corr = penguins.corr(numeric_only=True)

# Визуализируем их цветами
plt.figure(figsize=(7, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Корреляция признаков', fontsize=15)
plt.savefig('task7_exercise14.png', dpi=300)
plt.show()
```

# Упражнение 15 — Jointplot (совмещённый график: scatter + плотность)

```python
# Создаем jointplot
plot = sns.jointplot(
    data=penguins,  # Исправлено имя датасета
    x='bill_length_mm',
    y='bill_depth_mm',
    hue='species',  # Исправлен синтаксис
    kind='kde',
    fill=True
)

# Сохраняем через объект plot
plot.savefig('task7_exercise15.jpeg', dpi=85)  # Исправлено quality на dpi и имя файла
plt.show()
```

# Упражнение 16 — Попарные графики (pairplot)

```python
# Попарно сравниваем все числовые признаки у пингвинов
sns.pairplot(penguins, hue='species')
plt.show()
```
