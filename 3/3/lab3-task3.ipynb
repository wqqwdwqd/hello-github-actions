## Обработка строк с применением потокового редактора sed

```python
with open('src/books', 'w', encoding='utf-8') as f:
    f.write("Book one.\n")
    f.write("The second book.\n")
    f.write("The third.\n")
    f.write("This is book four.\n")
    f.write("Five.\n")
    f.write("This is the sixth.\n")
    f.write("This is book seven.\n")
    f.write("Eighth and last.\n")
print("Файл books создан!")
```

```python
# Создаём файл records
with open('src/records', 'w') as f:
    f.write("2,5 p\n")
print("Файл records создан!")

# Создаём файл appends
with open('src/appends', 'w') as f:
    f.write("3 a\\\n")
    f.write("My favorite book.\n")
print("Файл appends создан!")

# Создаём файл insert
with open('src/insert', 'w') as f:
    f.write("/This/ i\\\n")
    f.write("SKARBONKA.\n")
print("Файл insert создан!")
```
