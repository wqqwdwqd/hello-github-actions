# Приложение Voilà
## Визуализация данных

```python
import pandas as pd
import numpy as np
import plotly.express as px
from ipywidgets import interact, widgets
```

```python
# Создаём демонстрационные данные
np.random.seed(123)
dates = pd.date_range('2024-01-01', periods=100, freq='D')
sales_data = pd.DataFrame({
    'date': dates,
    'sales': np.random.randint(100, 1000, 100),
    'region': np.random.choice(['Север', 'Юг', 'Восток', 'Запад'], 100),
    'product': np.random.choice(['Товар A', 'Товар B', 'Товар C'], 100)
})

print("Данные о продажах:")
sales_data.head()
```

```python
# Создаём интерактивные элементы
region_selector = widgets.Dropdown(
    options=['Все регионы', 'Север', 'Юг', 'Восток', 'Запад'],
    value='Все регионы',
    description='Регион:'
)

product_selector = widgets.Dropdown(
    options=['Все товары', 'Товар A', 'Товар B', 'Товар C'],
    value='Все товары',
    description='Товар:'
)

@interact(region=region_selector, product=product_selector)
def update_plot(region, product):
    # Фильтруем данные
    filtered_data = sales_data.copy()
    
    if region != 'Все регионы':
        filtered_data = filtered_data[filtered_data['region'] == region]
    
    if product != 'Все товары':
        filtered_data = filtered_data[filtered_data['product'] == product]
    
    # Создаём график
    fig = px.line(filtered_data, x='date', y='sales', 
                  title=f'Продажи по дням ({region}, {product})',
                  color='region' if region == 'Все регионы' else None)
    
    # Показываем статистику
    total_sales = filtered_data['sales'].sum()
    avg_sales = filtered_data['sales'].mean()
    
    print(f"Общие продажи: {total_sales:,.0f}")
    print(f"Средние продажи в день: {avg_sales:.0f}")
    
    fig.show()
```

```python
# Сохраняем данные для Voilà
sales_data.to_csv('src/sales_data.csv', index=False)
print("Данные сохранены для Voilà приложения")
```
